## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Bagaimana kondisi cuaca memengaruhi rata-rata jumlah penyewaan sepeda selama periode 2011–2012?

- **Pertanyaan 2:** Bagaimana pola penyewaan sepeda pada setiap jam berbeda antara hari kerja dan hari libur selama periode 2011–2012?

### Cleaning Data

#### Fixing ... problem

```python
# Mengubah tipe data dteday menjadi datetime
day_df['dteday'] = pd.to_datetime(day_df['dteday'])
hour_df['dteday'] = pd.to_datetime(hour_df['dteday'])
```

```python
#Cek info data pada dataset day
day_df.info()
```

```python
# cek info data pada dataset hour
hour_df.info()
```

```python
# One hot encoding
season_labels = {
    1: "Spring",
    2: "Summer",
    3: "Fall",
    4: "Winter"
}

day_df["season"] = day_df["season"].map(season_labels)
hour_df["season"] = hour_df["season"].map(season_labels)
```

```python
workingday_labels = {
    0: "Holiday/Weekend",
    1: "Working Day"
}

day_df["workingday"] = day_df["workingday"].map(workingday_labels)
hour_df["workingday"] = hour_df["workingday"].map(workingday_labels)
```

```python
weekday_labels = {
    0: "Sunday",
    1: "Monday",
    2: "Tuesday",
    3: "Wednesday",
    4: "Thursday",
    5: "Friday",
    6: "Saturday"
}

day_df["weekday"] = day_df["weekday"].map(weekday_labels)
hour_df["weekday"] = hour_df["weekday"].map(weekday_labels)
```

### Pertanyaan 1: Bagaimana pengaruh kondisi cuaca terhadap rata-rata jumlah penyewaan sepeda selama periode tahun 2011–2012?

```python
#visualization 1
weather_avg = day_df.groupby("weathersit")["cnt"].mean().sort_values(ascending=False)

weather_avg

plt.figure(figsize=(8,5))

sns.barplot(
    x=weather_avg.index,
    y=weather_avg.values,
    palette="Blues_d"
)

plt.title("Average Bike Rentals by Weather Condition", fontsize=14, weight="bold")
plt.xlabel("Weather Condition")
plt.ylabel("Average Number of Rentals")

for index, value in enumerate(weather_avg.values):
    plt.text(index, value + 50, f"{value:.0f}", ha="center")

plt.show()
```

### Pertanyaan 2: Bagaimana pola penyewaan sepeda pada setiap jam berbeda antara hari kerja dan hari libur selama periode tahun 2011–2012?

```python
#visualization 2
hourly_pattern = (
    hour_df.groupby(["workingday", "hr"])["cnt"]
    .mean()
    .reset_index()
)

hourly_pattern.head()


plt.figure(figsize=(12,6))

sns.lineplot(
    data=hourly_pattern,
    x="hr",
    y="cnt",
    hue="workingday",
    linewidth=3
)

plt.title("Hourly Bike Rental Patterns: Working Day vs Holiday/Weekend",
          fontsize=14,
          weight="bold")

plt.xlabel("Hour")
plt.ylabel("Average Number of Rentals")
plt.xticks(range(0,24))

plt.legend(title="Day Type")

plt.show()
```

**Insight:** (Opsional)

Berdasarkan visualisasi, terdapat perbedaan pola penyewaan sepeda yang cukup jelas antara hari kerja (Working Day) dan hari libur/akhir pekan (Holiday/Weekend).

- Pada hari kerja, penyewaan sepeda menunjukkan dua puncak aktivitas, yaitu sekitar pukul 08.00 dan 17.00–18.00. Pola ini mengindikasikan bahwa sebagian besar pengguna memanfaatkan layanan bike sharing sebagai transportasi untuk berangkat dan pulang kerja (commuting).

- Sebaliknya, pada hari libur atau akhir pekan, jumlah penyewaan meningkat secara bertahap mulai pagi hari dan mencapai puncaknya sekitar pukul 13.00–15.00. Pola ini menunjukkan bahwa sepeda lebih banyak digunakan untuk aktivitas rekreasi atau perjalanan santai, bukan untuk mobilitas rutin.

- Selain itu, jumlah penyewaan pada malam hingga dini hari relatif rendah baik pada hari kerja maupun hari libur, yang mengindikasikan bahwa permintaan layanan bike sharing pada periode tersebut masih terbatas.

## Analisis Lanjutan (Opsional)

```python
#Correlation feature
plt.figure(figsize=(10,8))

sns.heatmap(
    day_df.select_dtypes(include="number").corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```

**Insight:** (Opsional)

Berdasarkan correlation matrix, terdapat beberapa hubungan yang cukup kuat antara variabel-variabel dalam dataset.

- Variabel registered memiliki korelasi positif yang sangat kuat dengan jumlah penyewaan (cnt) sebesar 0,95. Hal ini menunjukkan bahwa total penyewaan lebih banyak dipengaruhi oleh pengguna yang telah terdaftar dibandingkan pengguna kasual.

- Variabel casual juga memiliki korelasi positif dengan cnt sebesar 0,67, namun hubungan tersebut lebih rendah dibandingkan pengguna registered.

- Variabel temp dan atemp memiliki korelasi positif sedang terhadap jumlah penyewaan (cnt) sebesar 0,63. Ini mengindikasikan bahwa semakin tinggi suhu, jumlah penyewaan sepeda cenderung meningkat.

- Variabel hum (kelembapan) memiliki korelasi negatif yang lemah terhadap cnt sebesar -0,10, sedangkan windspeed memiliki korelasi negatif sebesar -0,23. Artinya, kelembapan dan kecepatan angin yang lebih tinggi cenderung diikuti oleh sedikit penurunan jumlah penyewaan, meskipun pengaruhnya tidak terlalu kuat.

- Korelasi yang sangat tinggi antara temp dan atemp (0,99) menunjukkan bahwa kedua variabel mengukur informasi yang hampir sama, sehingga pada analisis lanjutan seperti pemodelan prediktif, salah satu variabel dapat dipertimbangkan untuk dihilangkan guna mengurangi multikolinearitas.
